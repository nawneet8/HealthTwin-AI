# HealthTwin AI - 01 Data Exploration & EDA

This notebook performs Exploratory Data Analysis (EDA) on the synthetic EHR clinical records and wearable IoT continuous telemetry.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data_dir = '../data'
df_ehr = pd.read_csv(os.path.join(data_dir, 'ehr_data.csv'))
df_wearable = pd.read_csv(os.path.join(data_dir, 'wearable_data.csv'))

print('EHR Shape:', df_ehr.shape)
print('Wearable Shape:', df_wearable.shape)
df_ehr.head()
```

```python
# Demographics & Clinical Distribution
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.histplot(df_ehr['age'], ax=axes[0], kde=True, color='skyblue')
axes[0].set_title('Age Distribution')

sns.histplot(df_ehr['bmi'], ax=axes[1], kde=True, color='salmon')
axes[1].set_title('BMI Distribution')

sns.boxplot(x='diabetes_history', y='hba1c', data=df_ehr, ax=axes[2], palette='Set2')
axes[2].set_title('HbA1c by Diabetes History')
plt.tight_layout()
plt.show()
```

```python
# Continuous Glucose Trajectory Sample
sample_patient = df_wearable['patient_id'].iloc[0]
p_data = df_wearable[df_wearable['patient_id'] == sample_patient].copy()
p_data['timestamp'] = pd.to_datetime(p_data['timestamp'])

plt.figure(figsize=(14, 4))
plt.plot(p_data['timestamp'], p_data['continuous_glucose'], color='#0284c7', label='Glucose (mg/dL)')
plt.axhline(180, color='red', linestyle='--', label='Hyperglycemia (>180)')
plt.axhline(70, color='orange', linestyle='--', label='Hypoglycemia (<70)')
plt.title(f'Continuous Glucose Monitoring (CGM) - Patient {sample_patient}')
plt.xlabel('Timestamp')
plt.ylabel('Glucose (mg/dL)')
plt.legend()
plt.show()
```
